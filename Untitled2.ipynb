```python
def hitung_total(harga_per_kg, jumlah_kg):
    return harga_per_kg * jumlah_kg

def hitung_diskon(total, persen_diskon):
    potongan = total * (persen_diskon / 100)
    return total - potongan

total_awal = hitung_total(5000, 100)
total_akhir = hitung_diskon(total_awal, 10)
print(f"Total hasil panen setelah diskon: Rp{total_akhir}")
```
